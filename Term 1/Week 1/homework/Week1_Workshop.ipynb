### 2.2 Naming conventions

- What are the most important conventions for variable names in Python?
- Why is it important to stick to them?

Your answer: Use lowercase names with underscores between words (snake_case),
like `flight_co2` or `total_co2`. Make the name describe what the value is.
Constants are written in ALL_CAPS. Never use spaces, never start with a digit,
and avoid Python keywords and built-in function names such as `max`, `print` or
`type`. It is important to stick to these conventions because code is read far
more often than it is written: consistent, descriptive names make your code
understandable for other people and for your future self - and using a built-in
name can even break Python, as the `max` example shows.

### 3.3 if / elif / else

*Slide 15 - 20 minutes*

Many AI systems make decisions with exactly the structure you are about to
write. The same logic that flags a "high-risk loan applicant" can reproduce
human bias.

Python checks the conditions from top to bottom and runs the first one that is
True. Indentation is not styling in Python - it is the syntax that decides
which lines belong to which branch.

- Create the variable `income` with `income = int(input('Annual income in euros: '))`
- Write an `if` / `elif` / `else` block that assigns a risk level:
  - `income` below 20000 gives `'high risk'`
  - `income` below 40000 gives `'medium risk'`
  - anything else gives `'low risk'`
- Print: `Based on income alone, this person is classified as: [risk level]`
- Then also ask for `age`, and combine the two with `and` or `or` so that the
  classification uses both.

```python
income = int(input('Annual income in euros: '))

# Classification based on income alone:
if income < 20000:
    risk = 'high risk'
elif income < 40000:
    risk = 'medium risk'
else:
    risk = 'low risk'

print('Based on income alone, this person is classified as: ' + risk)

# Now also ask for age and combine the two with and / or:
age = int(input('Age: '))

if income < 20000 or age < 21:
    risk = 'high risk'
elif income < 40000 and age < 65:
    risk = 'medium risk'
else:
    risk = 'low risk'

print('Based on income and age, this person is classified as: ' + risk)
```

### 3.4 Reflection

- Is it fair to classify people on income alone? Which information is missing
from this model, and who would be harmed if a real organisation used it?

Your answer: No, income alone is not a fair basis to judge people. It says
nothing about someone's debts, family size, cost of living, health or job
stability, and it ignores why incomes differ - people are paid unequally for
reasons that have nothing to do with how reliable they are. A model like this
would reproduce that bias and harm people with low incomes (for example
students, caregivers and people in low-paid but stable jobs) by denying them
things they could actually pay for. A fairer model would need much more
context, transparent rules, and a way for people to appeal.

## 4. Green or Not?

Everything from the three waves, combined into one program. Individual work. This one goes into your portfolio.

In this assignment you combine everything from today into one program: input,
type conversion, calculation, and a decision. This is a portfolio piece - polish
it at home if you do not finish it in class.

**Data:**

Return short-haul flight: 400 kg CO2 each
Driving: 0.23 kg CO2 per km
Meat-heavy diet: 1400 kg CO2 per year
Sustainable annual budget: 2500 kg CO2

**Classification thresholds:**

Below 2500 kg: under budget
2500 to 4000 kg: slightly over budget
Above 4000 kg: far over budget

**Assignment:**
- Make variables for the four reference values above.
- Ask the user how many return flights they take per year, how many kilometres
  they drive per year, and whether they eat meat daily. Convert each answer to
  the right type.
- Calculate the total annual CO2 and assign it to a variable.
- Use `if` / `elif` / `else` to classify the total against the thresholds.
- Print the type of one of your variables.
- Print a personalised message containing both the total and the classification,
  for example: `You emit 5150 kg CO2 per year. That is far over budget.`

```python
# reference values:
flight_co2 = 400           # kg CO2 per return short-haul flight
car_co2_per_km = 0.23      # kg CO2 per kilometre driven
meat_diet_co2 = 1400       # kg CO2 per year for a meat-heavy diet
sustainable_budget = 2500  # sustainable annual CO2 budget in kg

# questions to the user:
flights = int(input('How many return flights do you take per year? '))
km_driven = int(input('How many kilometres do you drive per year? '))
meat_daily = input('Do you eat meat daily? (yes/no) ').lower() == 'yes'

# calculation:
total = flights * flight_co2 + km_driven * car_co2_per_km
if meat_daily:
    total = total + meat_diet_co2

# classification:
if total < sustainable_budget:      # below 2500 kg
    verdict = 'under budget'
elif total <= 4000:                 # 2500 to 4000 kg
    verdict = 'slightly over budget'
else:                               # above 4000 kg
    verdict = 'far over budget'

# print the result:
print(type(meat_daily))
print('You emit ' + str(round(total)) + ' kg CO2 per year. That is ' + verdict + '.')
```

Your answer: Before a government could use this calculator for a climate tax,
the model would have to be far more complete and transparent: the emission
numbers would have to come from published, regularly updated scientific data;
it would have to cover all major emission sources and not just three; people
would need to be able to check and challenge their result; and there would have
to be exceptions for people who genuinely cannot avoid driving or eating meat.
It would also have to be progressive, so that it does not hit the people who
pollute the least the hardest.
