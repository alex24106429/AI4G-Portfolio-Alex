 ## Learning Wave 1 — The String Toolkit



 ### 🧠 First: what *is* a string, really?



 A **string** is just a **row of characters** — exactly like a list, but every slot holds one character. So everything you already know about lists still works:



 - `text[0]` is the first character, `text[-1]` the last

 - `len(text)` is how many characters there are

 - `"bike" in text` checks whether that word appears



 What strings add is a **toolbox of methods** for cleaning and reshaping text. The ones you will use constantly:



 | Method | What it does |

 |---|---|

 | `.strip()` | remove spaces / newlines at the start and end |

 | `.lower()` / `.upper()` | change case, so `"Bike"` and `"bike"` count as the same word |

 | `.replace(old, new)` | swap one piece of text for another |

 | `.split()` | cut a sentence into a **list of words** |

 | `" ".join(words)` | glue a list of words back into one sentence |

 | `.count("x")` | how many times `"x"` appears |



 **The big idea for the whole week:** a computer cannot understand a sentence. The first move in *every* text and AI pipeline is to **clean** the text and **split it into pieces you can count.**

 ## Learning Wave 2 — From Text to Numbers



 One response is easy. But a city gets *thousands*. To find what people care about, you count words across **all** of them — this is called a **bag of words**, and it is the foundation of almost all classic NLP.



 The recipe reuses exactly what you already know:

 1. `for` over every response (Week 2)

 2. `.lower().split()` each response into words (Week 4)

 3. the **counting pattern** from Week 3: `counts[word] = counts.get(word, 0) + 1`



 ### Exercise 2: Which words come up most?

 **Steps:**

 1. Build a dictionary `counts` of how often each (lowercased) word appears, using the counting pattern.

 2. Print every word with its count using `.items()`.

 3. Find the **most mentioned** word with the comparison-variable pattern (Weeks 2–3 — no need for `max()` on a dict).

 4. **Bonus:** make a `stopwords` list (`"the"`, `"are"`, `"but"`, `"and"`, `"a"`, `"is"`, `"for"`, …) and skip those words with `continue`, so only meaningful words remain.

````python
responses = [
    "The bike lanes are great but too narrow",
    "More bike parking please the parking is always full",
    "Traffic lights are too short for pedestrians",
    "Great new park but we need more lights at night",
    "The park is lovely and the bike lanes are a great idea",
]

counts = {}

for response in responses:
    words = response.lower().split()
    for word in words:
        counts[word] = counts.get(word, 0) + 1

for word, count in counts.items():
    print(f"{word}: {count}")

most_mentioned_word = None
highest_count = -1

for word, count in counts.items():
    if count > highest_count:
        highest_count = count
        most_mentioned_word = word

print(f"\nMost mentioned word: '{most_mentioned_word}' with count {highest_count}")

stopwords = ["the", "are", "but", "and", "a", "is", "for", "too", "at", "we"]
filtered_counts = {}

for response in responses:
    words = response.lower().split()
    for word in words:
        if word in stopwords:
            continue
        filtered_counts[word] = filtered_counts.get(word, 0) + 1

print("\n--- Filtered Counts (Without Stopwords) ---")
for word, count in filtered_counts.items():
    print(f"{word}: ${word_count}")
````

 ## Learning Wave 3 — Reading & Writing Files

 So far the data lived *inside* your code. Real data lives in **files**. Python opens files with `open()`, and the safe way is the `with` block — it closes the file for you automatically:


 ```python

 with open("notes.txt", "w", encoding="utf-8") as f:
     f.write("first line\n")

 with open("notes.txt", "r", encoding="utf-8") as f:
     for line in f:
         print(line.strip())
 ```


 - `"w"` = write (overwrites the whole file), `"a"` = append, `"r"` = read.

 - Always pass `encoding="utf-8"` so accents and symbols don't break.

````python
# Big Assignment: Community Feedback Analyzer
# Design your own solution. It must: read a file, clean + split + count the text,
# and write a report. Build and test it one piece at a time.

import re

sample_feedback = [
    "The new bike lane on 4th Avenue is great and very safe for cyclists!\n",
    "ALERT: Broken glass and deep potholes on Main Street path near school 12.\n",
    "We need more street lights along Elm Road, it is too dark and feels unsafe at night.\n",
    "\n",
    "Bus route 7 is consistently 25 minutes late, service is terrible.\n",
    "DANGER: Pedestrian crossing signal at 5th Street is completely broken.\n",
    "Lovely new community garden, thank you for the wonderful green space!\n",
    "Speeding cars make the neighborhood unsafe for our 3 children.\n",
    "Great park renovation, excellent playground facilities for families.\n",
    "ALERT traffic lights are out at intersection 8, severe danger of accidents.\n",
]

feedback_input_file = "community_feedback.txt"
with open(feedback_input_file, "w", encoding="utf-8") as f:
    f.writelines(sample_feedback)

def read_feedback_file(filepath):
    """Reads non-empty responses from a file."""
    clean_lines = []
    with open(filepath, "r", encoding="utf-8") as f:
        for line in f:
            stripped = line.strip()
            if stripped:
                clean_lines.append(stripped)
    return clean_lines

def clean_word(word):
    """Removes common punctuation from word boundaries."""
    return word.strip(".,!?:;\"'()[]{}")

stopwords = {
    "the", "a", "an", "and", "or", "but", "in", "on", "at", "to", "for", 
    "of", "is", "are", "was", "were", "it", "we", "our", "too", "very", 
    "this", "with", "as", "from", "by"
}
urgent_keywords = ["unsafe", "broken", "alert", "danger"]
positive_words = ["great", "safe", "lovely", "wonderful", "excellent"]
negative_words = ["terrible", "dark", "broken", "unsafe", "late", "danger", "accidents"]

responses = read_feedback_file(feedback_input_file)
total_responses = len(responses)

total_words = 0
urgent_responses = []
word_frequencies = {}
all_numbers = []
pos_count = 0
neg_count = 0

for response in responses:
    response_lower = response.lower()
    if any(keyword in response_lower for keyword in urgent_keywords):
        urgent_responses.append(response)
        
    numbers_found = re.findall(r"\d+", response)
    all_numbers.extend(numbers_found)
    
    words = response_lower.split()
    total_words += len(words)
    
    for raw_word in words:
        cleaned = clean_word(raw_word)
        if not cleaned:
            continue
            
        if cleaned in positive_words:
            pos_count += 1
        if cleaned in negative_words:
            neg_count += 1
            
        if cleaned not in stopwords:
            word_frequencies[cleaned] = word_frequencies.get(cleaned, 0) + 1

avg_words_per_response = total_words / total_responses if total_responses > 0 else 0

sorted_words = sorted(word_frequencies.items(), key=lambda item: item[1], reverse=True)
top_5_keywords = sorted_words[:5]

def search_responses(term, feedback_list):
    term = term.lower()
    return [item for item in feedback_list if term in item.lower()]

search_term = "lights"
search_matches = search_responses(search_term, responses)

report_lines = [
    "=" * 64,
    "              COMMUNITY FEEDBACK ANALYSIS REPORT",
    "=" * 64,
    f"1. GENERAL METRICS:",
    f"   - Total responses analysed: {total_responses}",
    f"   - Total words counted: {total_words}",
    f"   - Average words per response: {avg_words_per_response:.2f}",
    "",
    "2. TOP 5 KEYWORDS (ignoring stopwords):",
]

for rank, (word, count) in enumerate(top_5_keywords, start=1):
    report_lines.append(f"   {rank}. '{word}' — {count} occurrences")

report_lines.extend([
    "",
    f"3. URGENT RESPONSES ({len(urgent_responses)} identified):",
])

for urgent_item in urgent_responses:
    report_lines.append(f"   [!] {urgent_item}")

report_lines.extend([
    "",
    "4. BONUS ANALYTICS:",
    f"   - Sentiment: {pos_count} positive vs {neg_count} negative words.",
    f"     Overall Community Tone: {'Positive' if pos_count > neg_count else 'Negative' if neg_count > pos_count else 'Neutral'}",
    f"   - Numbers/Routes referenced: {', '.join(all_numbers)}",
    f"   - Keyword Search ('{search_term}'): found in {len(search_matches)} response(s).",
    "=" * 64,
])

report_text = "\n".join(report_lines)

print(report_text)

with open("feedback_report.txt", "w", encoding="utf-8") as f:
    f.write(report_text + "\n")
````

 ## 5 Things to Remember



 1. **A string is a row of characters** — indexing, slicing, `len()` and `in` work just like with lists.

 2. **Clean before you count** — `.strip()`, `.lower()` and `.replace()` turn messy text into something comparable (`"Bike!"` and `"bike"` become the same).

 3. **`.split()` turns a sentence into a list of words; `" ".join()` turns it back** — this is *tokenisation*, the first step of every NLP system.

 4. **Counting words = the Week 3 counting pattern** — `counts[word] = counts.get(word, 0) + 1` is a "bag of words".

 5. **Files in, results out** — `with open(...) as f:` reads real data and writes reports safely. And remember: *the words missing from the file are missing from everything you build on it.*
